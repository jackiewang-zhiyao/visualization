```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(
    "applebyvariety_en.xlsx",
    sheet_name="Sheet1",
    header=2
)

df.head()
```

```python
print(df.columns)
```

```python
df = df[df["Variety"] != "Total"].copy()

df["Marketed Production ('000 lbs)"] = pd.to_numeric(
    df["Marketed Production ('000 lbs)"],
    errors="coerce"
)

df = df.dropna(
    subset=["Variety", "Marketed Production ('000 lbs)"]
)
```

```python
plt.figure(figsize=(14, 6))

plt.bar(
    df["Variety"],
    df["Marketed Production ('000 lbs)"])

max_position = df["Marketed Production ('000 lbs)"].argmax()
max_value = df["Marketed Production ('000 lbs)"].max()

plt.text(max_position,max_value * 1.02, "★",ha="center",va="bottom",fontsize=20)

plt.title("Ontario Apple Production by Variety",pad=15)

plt.xlabel("Apple Variety")
plt.ylabel("Marketed Production ('000 lbs)")

plt.xticks(rotation=45, ha="right")

plt.ylim(0, max_value * 1.12)

plt.tight_layout()
plt.show()
```
