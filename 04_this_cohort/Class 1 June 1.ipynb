```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'green','size':14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc='center')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
#loc can be 'left', 'center', 'right' for title, default is 'center'
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y,marker='*', color='indigo')
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', color='indigo', linestyle='--', linewidth=2)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*', color='#7424b9', linestyle='--', linewidth=2, markeredgecolor='#fa9359', markerfacecolor= '#000000')
```

```python
fig,ax = plt.subplots(figsize=(5, 3))
ax.grid(axis='y', color='blue', linestyle='-.', linewidth=2)
ax.plot(x,y,marker='*', color='#7424b9', linestyle='--', linewidth=2, markeredgecolor='#fa9359', markerfacecolor= '#000000')
```
